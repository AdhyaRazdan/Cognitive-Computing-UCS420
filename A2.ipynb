```python
print("Question 1")
num = 1024170337
L = [int(digit) * 10 for digit in str(num)]
print("Original is ", L)

L.append(76)
print("After appending ", L)
L.insert(4,1)
print("After inserting ", L)

L.remove(1)
print("After remove ", L)
L.pop()
print("After pop ", L)

L.sort()
print("After sorting in ascending order ", L)
L.sort(reverse=True)
print("After sorting in descending order ", L)

print("First three ", L[:3])
print("Last three ", L[-3:])

avg = sum(L) / len(L)
print("Average ", avg)

greater_than_average = [x for x in L if x > avg]
print(greater_than_average)
```

```python
print("Question 3")
import random
random.seed(1024170337)

numbers = [random.randint(100, 900) for _ in range(100)]

print("Random numbers:")
print(numbers)

odd_numbers = [num for num in numbers if num % 2 != 0]
print("Odd numbers " ,odd_numbers)
print("Number of odd numbers ", len(odd_numbers))

even_numbers = [num for num in numbers if num % 2 == 0]
print("Even numbers ", even_numbers)
print("Number of even numbers:", len(even_numbers))

def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True

prime_numbers = [num for num in numbers if is_prime(num)]

print("Prime numbers ", prime_numbers)
print("Number of prime numbers:", len(prime_numbers))

most_frequent = max(set(numbers), key=numbers.count)
frequency = numbers.count(most_frequent)

print("\nMost frequently occurring number:", most_frequent)
print("Number of times it occurs:", frequency)
```

```python
print("Question 4")
digits = [1, 0, 2, 4, 1, 7, 0, 3]

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}
print("Set A:", A)
print("Set B:", B)

union = A.union(B)
print("Union of A and B:", union)

intersection = A.intersection(B)
print("Intersection of A and B:", intersection)

A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)

symmetric_diff = A.symmetric_difference(B)
print("Symmetric difference:", symmetric_diff)

print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

X = input("Enter a value X: ")
A.discard(X)


print("Set A after discard:", A)
```

```python
print("Question 5")

my_dict = {
    "name": "Adhya Razdan",
    "roll_no": "1024170337",
    "branch": "COPC",
    "age": 20,
    "city": "Patiala"
}

print("Original dictionary:")
print(my_dict)

my_dict["location"] = my_dict.pop("city")
print("After renaming city ", my_dict)

my_dict["cgpa"] = 8.84
print("After adding CGPA ", my_dict)

my_dict["age"] = my_dict["age"] + 1

print("After increasing age by 1 ", my_dict)

dict_pop = my_dict.copy()
removed_branch = dict_pop.pop("branch")

print("Dictionary after using pop() ", dict_pop)
print("Value removed by pop():", removed_branch)

dict_del = my_dict.copy()
del dict_del["branch"]

print("Dictionary after using del ", dict_del)

print("Key-value pairs ")

for key, value in my_dict.items():
    print(key, "→", value)

if "email" in my_dict:
    print("\nEmail:", my_dict["email"])
else:
    print("\nEmail: Not available")

friend_dict = {
    "name": "Riya",
    "roll_no": "1024170317",
    "branch": "COPC",
    "age": 20,
    "city": "Bathinda"
}

merged_dict = {**my_dict, **friend_dict}

print("Merged dictionary ", merged_dict)

string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("Dictionary containing only string values ", string_values)
```
