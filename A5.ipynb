```python
x = np.array([1,2,3,4,5,1,2,1,1,1]) 
values, counts = np.unique(x, return_counts=True)
max_count = np.max(counts)
most_frequent = values[counts == max_count]
print("Most frequent value = ", most_frequent)
print("Count of most frequent element = ", max_count)

for value in most_frequent:
    indices = np.where(x == value)[0]
    print("Indices of", value, ":", indices)

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])
values, counts = np.unique(y, return_counts=True)
max_count = np.max(counts)
most_frequent = values[counts == max_count]
print("Most frequent value = ", most_frequent)
print("Count of most frequent element = ", max_count)

for value in most_frequent:
    indices = np.where(y == value)[0]
    print("Indices of", value, ":", indices)
```

```python
start = 10
stop = 100
num = 25
endpoint = True
Adhya = np.linspace(start, stop, num, endpoint)
print(Adhya)
```

```python
print("Dimensions:", Adhya.ndim)
print("Shape:", Adhya.shape)
print("Total elements:", Adhya.size)
print("Data type:", Adhya.dtype)
print("Total bytes:", Adhya.nbytes)
print("Reshaped array: ")
Adhya.reshape(5,5) # accepts only 5,5 
print(Adhya.T)
```

```python
ucs420_Adhya = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_Adhya)

print("Mean:", np.mean(ucs420_Adhya))
print("Median:", np.median(ucs420_Adhya))
print("Maximum:", np.max(ucs420_Adhya))
print("Minimum:", np.min(ucs420_Adhya))
print("Unique elements:", np.unique(ucs420_Adhya))

reshaped_ucs420_Adhya = ucs420_Adhya.reshape(4, 3)

print("\nReshaped 4 x 3:")
print(reshaped_ucs420_Adhya)

resized_ucs420_Adhya = np.resize(ucs420_Adhya, (2, 3))

print("\nResized 2 x 3:")
print(resized_ucs420_Adhya)
```
